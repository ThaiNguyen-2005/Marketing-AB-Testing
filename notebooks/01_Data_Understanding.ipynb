# DATA UNDERSTANDING

The purpose of this notebook is to understand the structure and quality of the marketing campaign dataset before performing statistical analysis and machine learning.

This stage includes:

* Data loading
* Data cleaning
* Missing value analysis
* Duplicate analysis
* Data type checking
* Initial feature engineering
* Dataset merging


```python
# import thư viện
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
# file control_group.csv 
control_df = pd.read_csv('../data/raw/control_group.csv', sep=';')
control_df.head()
```

```python
# file test_group.csv
test_df = pd.read_csv('../data/raw/test_group.csv', sep=';')
test_df.head()
```

```python
control_df.isnull().sum()
```

```python
test_df.isnull().sum()
```

```python
# kiểm tra kích thước của hai DataSet
print(control_df.shape)
print(test_df.shape)
```

```python
control_df.info()
test_df.info()
```

```python
control_df.describe()
test_df.describe()
```

```python
new_columns = [
    'campaign_name',
    'date',
    'spend_usd',
    'impressions',
    'reach',
    'website_clicks',
    'searches',
    'view_content',
    'add_to_cart',
    'purchase'
]

control_df.columns = new_columns
test_df.columns = new_columns
```

```python
control_df.head()
```

```python
test_df.head()
```

```python
control_df['group'] = 'control'
test_df['group'] = 'test'
```

```python
control_df.head()
```

```python
test_df.head()
```

```python
df = pd.concat([control_df, test_df])
```

```python
df.reset_index(drop=True, inplace=True)
```

```python
df.head()
```

```python
print("Final Dataset Shape:", df.shape)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['date'] = pd.to_datetime(
    df['date'],
    format='%d.%m.%Y'
)
```
