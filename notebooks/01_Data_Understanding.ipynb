### Xử lý Missing Value (Dữ liệu khuyết thiếu)

Qua kiểm tra, ta thấy có 1 ngày duy nhất (`2019-08-05`) ở nhóm Control bị khuyết thiếu (NaN) trên 7 cột chỉ số chuyển đổi quan trọng (impressions, reach, website_clicks, searches, view_content, add_to_cart, purchase) trong khi cột spend_usd vẫn ghi nhận chi phí (1835 USD).

**Phương án xử lý:**
- Do số lượng bản ghi nhỏ (30 ngày mỗi nhóm), việc giữ lại dòng bị khuyết này sẽ gây lỗi `NaN` khi thực hiện các phép kiểm định thống kê hoặc hồi quy.
- Phương án điền khuyết (imputation) bằng giá trị trung bình/trung vị có thể làm giảm phương sai tự nhiên và ảnh hưởng đến kết quả phân tích A/B test.
- Do đó, ta quyết định **loại bỏ (drop)** dòng dữ liệu khuyết thiếu này ra khỏi tập dữ liệu sạch để đảm bảo các bước phân tích thống kê tiếp theo diễn ra chính xác và không bị lỗi.

```python
# Loại bỏ các dòng chứa giá trị null (chỉ có 1 dòng duy nhất vào ngày 2019-08-05 ở nhóm control)
df_clean = df.dropna().reset_index(drop=True)
print(f"Kích thước ban đầu: {df.shape}, Sau khi làm sạch null: {df_clean.shape}")

# Save at processed data
df_clean.to_csv(
    '../data/processed/cleaned_marketing.csv',
    index=False
)
```
